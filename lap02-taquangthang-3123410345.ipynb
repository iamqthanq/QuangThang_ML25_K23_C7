### Bài 1. Tính giá trị hàm số
**Yêu cầu**: Hãy viết chương trình nhập vào số thực x và tính giá trị biểu thức:
$$
    f\left(x\right) = x + \frac{x^5}{5!} + \frac{\sqrt{|x|}}{x^{3/2}}
$$
> Moi ban nhap vao gia tri cua bien so x: **2.3** <br/>
> Gia tri cua ham so f(**2.3**) = **3.27**.

### Bài làm


````python
# Khai báo thư viện
import math

# Nhập dữ liệu
x   = float(input('Moi ban nhap vao gia tri cua bien so x: '))
# Xử lý
f_x = x + (x**5) / math.factorial(5) + math.sqrt(abs(x)) / (x**(3/2))

# Xuất dữ liệu
print(f'Gia tri cua ham so f({x}) = {f_x: .2f}.')
````

### Bài 7. Giải phương trình bậc 2
Nhập 3 số thực a, b, c (có kiểm tra dữ liệu nhập số thực). Hãy xây dựng hàm giải phương trình bậc 2 $ax^2+bx+c=0$ và in kết quả giải ra màn hình.
> Moi ban nhap he so a, b, c:  **a b c**<br/>
> Ban nhap sai! Moi ban nhap lai!<br/>
> Moi ban nhap he so a, b, c:  **1 a 3**<br/>
> Ban nhap sai! Moi ban nhap lai!<br/>
> Moi ban nhap he so a, b, c: **1 5 6**<br/>
> Phuong trinh bac 2 **1x^2 + 5x + 6 = 0** co **2**  nghiem x1 = **-2.00**, x2 = **-3.00**.

### Bài làm

````python
def NhapDuLieu():
    """
    Input: 
        Ban phim = 1 5 6
    Output:
        (a, b, c) --> (1, 5, 6)
    """
    while True:
        try:
            data = input("Moi ban nhap he so a, b, c: ").split()
            if len(data) != 3:
                raise ValueError
            a, b, c = map(float, data)
            return a, b, c
        except:
            print("Ban nhap sai! Moi ban nhap lai!") 
    

    return a, b, c
# NhapDuLieu
````

````python
# Kiểm tra hàm
a, b, c = NhapDuLieu()
````

````python
def GiaiPhuongTrinhBac2(a, b, c):
    """
    Input: a, b, c
    Output: 
    + flag = -1 (VSN), 0 (VN), k (k nghiem)
    + () --> flag = -1, 0
    + (x) --> flag = 1
    + (x1, x2) --> flag = 2
    """
    import math
    flag = None
    x = ()
    
    if a == 0:
        if b == 0:
            if c == 0:
                flag = -1  # vô số nghiệm
            else:
                flag = 0   # vô nghiệm
        else:
            # phương trình bậc nhất: bx + c = 0
            x = (-c / b,)
            flag = 1
    else:
        delta = b**2 - 4*a*c
        if delta < 0:
            flag = 0
        elif delta == 0:
            x = (-b / (2*a),)
            flag = 1
        else:
            x1 = (-b + math.sqrt(delta)) / (2*a)
            x2 = (-b - math.sqrt(delta)) / (2*a)
            x = (x1, x2)
            flag = 2
    # ...    
    
    return flag, x
# GiaiPhuongTrinhBac2
````

````python
print("a = 0")
flag, x = GiaiPhuongTrinhBac2(0, 0, 0)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 0, 5)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')

print("a <> 0")
flag, x = GiaiPhuongTrinhBac2(5, 1, 2)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 2, 1)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')
````

````python
import random
def SinhNgauNhien(n, vmin = -10, vmax = 10):
    """
    Input:
        n = 30, vmin = -10, vmax = 10
    Output:
        a = [-1, -4, -8, 6, 4, -5, -1, -4, -9, -5, -10, 0, -6, 2, 1, -9, -9, 10, -8, -9, -7, 9, 2, -4, -1, -6, 5, -6, 9, -8]
    """    
    a = []
    a = [random.randint(vmin, vmax) for _ in range(n)]
    
    return a
# SinhNgauNhien
````

````python
def XuLyKhongTrung(a):
    """
    Input:
        a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
        s = ['A', 'C', 'F', 'G', 'H', 'T']
    """
    s = set([])
    s = set(a) 
    # ...
        
    return sorted(s)
# XuLyKhongTrung
````

````python
def DemSoLanXuatHien(a):
    """
    Input:
        a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
        dem = {'A': 2, 'G': 1, 'C': 2, 'F': 2, 'T': 1, 'H': 1}
    """
    dem = {} # (k, v): k (ky tu) v (so lan xuat hien)
    for x in a:
        if x in dem:
            dem[x] += 1
        else:
            dem[x] = 1
    # ...    
    
    return dem
# DemSoLanXuatHien
````

````python
a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
b = DemSoLanXuatHien(a)
print(b)
````

#### 4. Giải hệ phương trình tuyến tính
Giải hệ phương trình tuyến tính sau:
$$
\left\{\begin{matrix}
        4x_1  + 3x_2 - 5x_3 = 2 \\ 
        -2x_1 - 4x_2 + 5x_3 = 5 \\ 
        8x_1 + 8x_2 = -3 \\ 
\end{matrix}\right.
$$

````python
import numpy as np




A = np.array([[4, 3, -5],
              [-2, -4, 5],
              [8, 8, 0]])
y = np.array([2, 5, -3])

x = np.linalg.solve(A, y)

yy = A @ x
diff = np.linalg.norm(yy - y, 2)
# ...

print(f"+ Ma tran A: \n{A}")
print(f"+ Vector y: {y}")
print(f"+ Nghiem x: {x}")
print(f"+ Kiem tra: yy = Ax = {yy} ==> ||yy -y ||_2 = {diff: .2f}")
````

### Bài 3. Tính gần đúng số pi dùng phương pháp Monte Carlo

Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông. Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình vuông như sau:
$$
\frac{N_C}{N} \approx  \frac{\mathbb{S}_C}{\mathbb{S}_N}
$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $\mathbb{S}_C$ và $\mathbb{S}_N$ lần lượt là diện tích của hình tròn và hình vuông.
$$ \mathbb{S}_C = \pi * r^2 $$
$$ \mathbb{S}_N = (2*r)^2 = 4r^2 $$
Do đó, công thức trên trở thành:
$$
\frac{N_C}{N} \approx \frac{\pi * r^2}{4r^2} = \frac{\pi}{4}
$$
Suy ra, ta có giá trị số $\pi$ được tính như sau:
$$
\pi \approx \frac{4N_C}{N}
$$

Các bạn hãy lập trình tính sấp xỉ số $\pi$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$, $N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.

**Hướng dẫn**: Viết chương trình phát sinh $N$ điểm $(x_i, y_i)$ với $x_i, y_i \in [-1, -1]$. $(x_i, y_i)$ thuộc hình tròn tâm $(0, 0)$ bán kính 1 nội tiếp hình vuông có tâm tại $(0, 0)$ độ dài cạnh 2 khi và chỉ khi $x_i^2 + y_i^2 <= 1$.

````python
import matplotlib.pyplot as plt # khai báo thư viện vẽ pyplot
import math

circle = plt.Circle((0, 0), 1, color='g') # tạo đối tượng vòng tròn

fig, ax = plt.subplots(figsize=(6, 6)) # lấy figure và vùng vẽ

plt.plot(0, 0, 'o', color=(0.9, 0.9, 1.0), alpha=0.8) # vẽ điểm tại tâm (0,0)
ax.add_patch(circle) # thêm vòng tròn vào vùng vẽ

x = 0.75; y = math.sqrt(1 - (x ** 2)) # (x = 0.75, y = căn bậc 2 của 1 - x^2)
plt.arrow(0, 0, x, y)

plt.xlim(-1, 1) # giới hạn trục x
plt.ylim(-1, 1) # giới hạn trục y

plt.savefig('pi.png') # lưu đồ thị xuống tập tin
plt.show()            # hiển thị đồ thị
````

````python
import numpy as np

def calc_pi_monte_carlo(n = 100):
    pi = 0
    
    # gợi ý: sử dụng np.random.rand(s1, s2, ...) --> sinh ngẫu nhiên các số trong nửa đoạn [0.0, 1.0)
    x = np.random.rand(n) * 2 - 1   # [0,1) → [-1,1)
    y = np.random.rand(n) * 2 - 1

    inside_circle = np.sum(x**2 + y**2 <= 1)

    pi = 4 * inside_circle / n   
    # ...
    
    return pi
# calc_pi_monte_carlo
````

````python
import math

print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
````
